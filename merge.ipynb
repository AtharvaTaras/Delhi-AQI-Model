```python
import pandas as pd
```

```python
wind_df = pd.read_csv(r'data\punjab_wind_2021_2025.csv')
fire_df = pd.read_csv(r'data\punjab_fires_2021_2025.csv')
weather_df = pd.read_csv(r'data\delhi_weather_2021_2024.csv')
aq_df = pd.read_csv(r'data\delhi_aqi_2021_2024.csv')
```

```python
wind_df.head()
```

```python
weather_df["DATE"] = pd.to_datetime(weather_df["DATE"]).dt.strftime("%Y-%m-%d")
weather_df.head()
```

```python
merge_1 = pd.merge(wind_df, fire_df, how='inner', on='acq_date')
merge_1.head()
```

```python
merge_2 = pd.merge(merge_1, weather_df, how='inner', left_on='acq_date', right_on='DATE')
merge_2.drop(columns=['Data Acquisition Time', 'DATE'], inplace=True)
merge_2.head()
```
