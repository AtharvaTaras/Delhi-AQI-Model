```python
import pandas as pd
```

```python
wind_df = pd.read_csv(r'data\punjab_wind_2021_2025.csv')
fire_df = pd.read_csv(r'data\punjab_fires_2021_2025.csv')
weather_df = pd.read_csv(r'data\delhi_weather_2021_2024.csv')
aq_df = pd.read_csv(r'data\delhi_aqi_2021_2024.csv')
```

```python
wind_df.head()
```

```python
weather_df["DATE"] = pd.to_datetime(weather_df["DATE"]).dt.strftime("%Y-%m-%d")
weather_df.head()
```

```python
merge_1 = pd.merge(wind_df, fire_df, how='inner', on='acq_date')
merge_1.head()
```

```python
merge_2 = pd.merge(merge_1, weather_df, how='inner', left_on='acq_date', right_on='DATE')
merge_2.drop(columns=['Data Acquisition Time', 'DATE'], inplace=True)
merge_2.head()
```

```python
merge_2[["acq_date"]].head(10)
merge_2["acq_date"].dtype
```

```python
merge_2.acq_date.head()
```

```python
merge_2["acq_date"] = pd.to_datetime(
    merge_2["acq_date"], format="%Y-%m-%d", errors="coerce"
)
```

```python
len(fire_df), len(merge_2)
```

```python
fire_df.head()
```

```python
merge_2.to_csv(r'data\telemetry_weather_2021-2025.csv', index=False)
```

```python
# numeric columns
daily_numeric = merge_2.groupby("acq_date").mean(numeric_only=True)

# non-numeric columns (take first value of the day)
daily_non_numeric = merge_2.groupby("acq_date").agg(lambda x: x.iloc[0])
```

```python
for col in daily_numeric.columns:
    fire_df[col] = fire_df["acq_date"].map(daily_numeric[col])

for col in daily_non_numeric.columns:
    fire_df[col] = fire_df["acq_date"].map(daily_non_numeric[col])
```

```python
# Merge AQI into full merged dataset (preserve all rows from big_df)
# Read datasets with parsed dates
big_df = pd.read_csv(r'data\full_merged.csv', parse_dates=['acq_date'])
aq_df = pd.read_csv(r'data\delhi_aqi_2021_2024.csv', parse_dates=['acq_date'])

# Ensure date columns are datetime
big_df['acq_date'] = pd.to_datetime(big_df['acq_date'], errors='coerce')
aq_df['acq_date'] = pd.to_datetime(aq_df['acq_date'], errors='coerce')

# Left join to preserve all rows in big_df
big_df_with_aqi = big_df.merge(aq_df[['acq_date', 'AQI']], on='acq_date', how='left')

print('big_df rows:', len(big_df), 'merged rows:', len(big_df_with_aqi))
print('Missing AQI after merge:', big_df_with_aqi['AQI'].isna().sum())

# Save merged file (do not overwrite original unless you want to)
big_df_with_aqi.to_csv(r'data\full_merged_with_aqi.csv', index=False)
print('Saved merged file to data/full_merged_with_aqi.csv')
```
